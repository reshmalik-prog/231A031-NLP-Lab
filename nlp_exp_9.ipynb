```python
#Name:khan alfia shamsul uin:231A025
import nltk
from nltk.corpus import wordnet as wn
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    tokens = nltk.word_tokenize(text.lower())
    processed_tokens = [
        lemmatizer.lemmatize(token)
        for token in tokens
        if token.isalpha() and token not in stop_words
    ]
    return set(processed_tokens)

def lesk(word , sentence):
    best_sense = None
    max_overlap = 0
    sentence_tokens = preprocess_text(sentence)

    for sense in wn.synsets(word):
        definition_tokens = preprocess_text(sense.definition())
        example_tokens = preprocess_text(' '.join(sense.examples()))

        context_tokens = definition_tokens.union(example_tokens)
        overlap = len(sentence_tokens.intersection(context_tokens))

        if overlap > max_overlap:
            max_overlap = overlap
            best_sense = sense

    return best_sense
```
