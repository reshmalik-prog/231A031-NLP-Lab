```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.util import bigrams
from nltk.probability import FreqDist
```

```python
print("MALIK RESHMA 231A031")
# Step 1: Load and preprocess the text corpus

def preprocess(text) :
    # Convert to Lowercase
    text = text.lower()

    # Tokenize the text
    tokens = word_tokenize(text)

    # Remove punctuation and stopwords
    stop_words = set(stopwords.words('english'))
    tokens = [token for token in tokens if token. isalnum() and token not in stop_words]
    return tokens
```

```python
print("MALIK RESHMA 231A031")

# Sample text corpus
# corpus = "The quick brown fox jumps over the lazy dog. The fox is quick and the dog is lazy."
corpus = "Joe made the sugar cookies; Susan decorated them"

# Step 2: Preprocess the text
tokens = preprocess(corpus)

print(tokens)
```

```python
print("MALIK RESHMA 231A031")
# Step 3: Generate bigrams from the tokenized text
bigrams_list = list(bigrams(tokens))
print(bigrams_list)
```

```python
print("MALIK RESHMA 231A031")

# Step 4: Count bigram frequencies using FreqDist
bigram_freq = FreqDist(bigrams_list)
bigram_freq
```

```python
print("MALIK RESHMA 231A031")

# Step 5: Calculate conditional frequencies to predict the next word
from nltk.probability import ConditionalFreqDist

cfd = ConditionalFreqDist(bigrams_list)
print(cfd.items())

```

```python
print("MALIK RESHMA 231A031")
#create trigrams

tri_grams = list(trigrams(words))
print(tri_grams[:10000])
```

```python
print("MALIK RESHMA 231A031")
# Build a trigram model
model = defaultdict(lambda: defaultdict(lambda: 0))
```

A nested defaultdict is created to hold the trigram model. It maps each pair of words (w1,
w2) to a dictionary that contains counts of all possible third words (w3) that follow
them.

Counting Co-Occurrence Frequencies:

```python
print("MALIK RESHMA 231A031")

# Count frequency of co-occurrence
for w1, w2, w3 in tri_grams:
  model[(w1, w2)][w3] += 1

```

Transforming Counts into Probabilities


```python
print("MALIK RESHMA 231A031")

# Transform the counts into probabilities
for w1,w2 in model:
  total_count = float(sum(model[w1,w2].values()))
for w3 in model[w1,w2]:
  model[w1,w2] [w3] /= total_count

```

Function to Predict the Next Word:


```python
print("MALIK RESHMA 231A031")

# Function to predict the next word
def predict_next_word(w1, w2):
  next_word = model[w1, w2]
  if next_word:
    predicted_word = max(next_word, key=next_word.get)
    # Choose the most Likely next word
    return predicted_word
  else:
    return "No prediction available"
```
