Perform the standard preprocessing steps:

* Convert text to lowercase
* Remove HTML, URLs, Emails
* Tokenize the text
* Remove punctuation


```python
#To convert text to lowercase

print("Reshma Malik", "UIN: 231A031\n")

print(text.lower())
```

```python
text_lower = text.lower()
```

To remove HTML, URL

```python
print("Malik Reshma", "UIN: 231A031\n")

import re
pattern=r'https?://www\.\S+|www\.\S+ | \S\@\S+'
text_url = re.sub(pattern," ", text_lower)
print(text_url)
```

```python
pattern=r'https?://www\.\S+ |www\.\S+|\S+\@\S+'
text_emails = re.sub(pattern," ", text_lower)
print(text_emails)
```

```python
pattern=r'\d+\S+'
text_percentage = re.sub(pattern," ", text_emails)
print(text_percentage)
```

```python
import emoji

text_emoji=emoji.replace_emoji(text_percentage, replace="")
print(text_emoji)
```
